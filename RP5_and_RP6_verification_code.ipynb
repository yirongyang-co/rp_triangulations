### Verify basic properties of $\mathcal{P}_{6,48}$ and $\mathcal{P}_{7,90}$
In particular, check that they are both simplicial and satisfy Condition (1) (the disjoint star condition). Compute their $f$-vectors, the order of their automorphism groups, and the degree sequences of their graphs. Check also that the graph of $\mathcal{P}_{6,48}$ is vertex transitive. 

```python
def check_width(P):
    
    """
    input: a cs simplicial polytope P
    output: True if no pair of antipodes share a common neighbor
            False and the first violating vertex and common neighbors otherwise
    """
    
    Vert = P.vertices()
    V = [tuple(v) for v in Vert]
    
    G = P.graph()

    # dictionary recording antipodal pairs
    antipode = {}
    for v in V:
        anti = tuple(-c for c in v)
        antipode[v] = anti

    # check for shared neighbors
    for v in V:
        a = antipode[v]
        neighbors_v = {tuple(n) for n in G.neighbors(Vert[V.index(v)])}
        neighbors_a = {tuple(n) for n in G.neighbors(Vert[V.index(a)])}
        common = neighbors_v.intersection(neighbors_a)
        if common:
            print("Violation at vertex", v)
            print("Common neighbors:", common)
            return False

    return True
```

```python
polytopes = {"P_6_48": P_6_48_verts, "P_7_90": P_7_90_verts}

for name, P_verts in polytopes.items():
    print("-----------------------------------------------")
    print("Polytope: ", name)
    P = Polyhedron(vertices=P_verts, base_ring = QQ) # take the convex hull of the vertices
    G = P.graph() # the graph (1-skeleton) of the polytope

    print("Simplicial? ", P.is_simplicial())
    print("f-vector: ", P.f_vector())
    print("Condition (1) satisfied? ", check_width(P))
    print("Order of combinatorial automorphism group: ", P.boundary_complex().automorphism_group().order())
    print("Vertex transitive graph? ", G.is_vertex_transitive())
    print("Graph degree sequence: ", G.degree_sequence())
```

### Verify the pattern of the facets of $\mathcal{P}_{6,48}$
In particular, check that every facet is the convex hull (combinatorially, the join) of faces from the different cubes in $\{V_{\{1,2,3\}},V_{\{1,4,5\}},V_{\{1,4,6\}},V_{\{2,3,4\}},V_{\{2,5,6\}},V_{\{3,5,6\}\}}$, where $V_S$ denotes the set of vertices whose support is exactly the coordinates indexed by $S$.

```python
# for each facet, print out the cubes its vertices belong to
for facet in P_6_48.facets():
    cube_for_verts = ()
    for v in facet.vertices():
        S = {i+1 for i, x in enumerate(tuple(v)) if x != 0}
        cube_for_verts = cube_for_verts + (f"V_{S}",)
    print(cube_for_verts)
```

### Examine the inner products of every pair of vertices of $\mathcal{P}_{6,48}$
In particular, check that there is an edge between vertices $p$ and $q$ if and only if $\langle p,q\rangle>0$

```python
edge_set_poly = {frozenset(tuple(v) for v in edge.vertices()) 
                 for edge in P_6_48.face_generator(face_dimension=1)} # the edge set obtained directly from the polytope

edges_by_dot = {frozenset({p, q}) for i, p in enumerate(P_6_48_verts)
                             for j, q in enumerate(P_6_48_verts)
                             if j > i and vector(p).dot_product(vector(q)) > 0} # the edge set obtained from the dot product condition

print(edges_by_dot == edge_set_poly)
```

### Examine the links of $\mathcal{P}_{6,48}$
In particular, list out the number of distinct combinatorial types of the links of $i$-dimensional faces for $i \in {0, 1, 2, 3}$, and examine closely the links of $0$-faces and $2$-faces.

```python
iso_classes_of_links = {}


for i in range(4):
    iso_classes = []
    for face in P_6_48_bd.n_faces(i):  # iterate over all i-faces
        face_link = P_6_48_bd.link(face)  # get the link of the face
    
        # Check if this link is isomorphic to any already in iso_classes
        is_new = True
        for K in iso_classes:
            if face_link.is_isomorphic(K):
                is_new = False
                break
    
        # If it's new, append it to the list
        if is_new:
            iso_classes.append(face_link)

    iso_classes_of_links[i] = iso_classes
    
    print(f"Number of distinct combinatorial types of the links of {i}-faces: {len(iso_classes)}")
```
