```python
def check_width(P):
    
    """
    input: a cs simplicial polytope P
    output: True if no pair of antipodes share a common neighbor
            False and the first violating vertex and common neighbors otherwise
    """
    
    Vert = P.vertices()
    V = [tuple(v) for v in Vert]
    
    G = P.graph()

    # dictionary recording antipodal pairs
    antipode = {}
    for v in V:
        anti = tuple(-c for c in v)
        antipode[v] = anti

    # check for shared neighbors
    for v in V:
        a = antipode[v]
        neighbors_v = {tuple(n) for n in G.neighbors(Vert[V.index(v)])}
        neighbors_a = {tuple(n) for n in G.neighbors(Vert[V.index(a)])}
        common = neighbors_v.intersection(neighbors_a)
        if common:
            print("Violation at vertex", v)
            print("Common neighbors:", common)
            return False

    return True
```
